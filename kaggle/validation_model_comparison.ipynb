```python
PROTOCOL = {'commit': 'ea41d0b791a45e1e84426b9df93b133b15aa313c', 'source_checkpoint_sha256': 'a77dda823d02c2d900a5908972b1752be1d1dc4866a48cb032ded29273ea52c7', 'xy': {'1.5_meter_los_person': [1.5, 0], '1.5_meter_los_person_repeat': [1.5, 0], '1_human_in_los_0.5m': [0.5, 0], '1_meter_Los_person': [1.0, 0], '1m_off_LOS_left_0.5m': [1.0, -0.5], '1m_off_LOS_left_1m': [1.0, -1.0], '1m_off_LOS_right_0.5m': [1.0, 0.5], '1m_off_los_right_1m': [1.0, 1.0], '1m_off_los_right_1m_repeat': [1.0, 1.0], '2_meter_los_person': [2.0, 0], '2_meter_los_person_repeat': [2.0, 0], '2_walls_2.5m_1_person_behind ESPby1m': [-1.0, 0], '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': [4.5, 0], '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': [3.5, 0], '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los_repeat': [3.5, 0], '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': [1.25, 0], '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': [5.0, 0], '2_walls_4m_1_person_between_midpoint': [2.0, 0], '2_walls_4m_1_person_between_midpoint_repeat': [2.0, 0], '2m_off_Los_left_1m': [2.0, -1.0], '2m_off_Los_left_1m_repeat': [2.0, -1.0], '2m_off_Los_right_1m': [2.0, 1.0], '2m_off_los_left_0.5m': [2.0, -0.5], '2m_off_los_right_0.5m': [2.0, 0.5], 'no_wall_1_person_on_los_x0.75m': [0.75, 0], 'no_wall_1_person_on_los_x1.25m': [1.25, 0], 'no_wall_1_person_on_los_x1.75m': [1.75, 0], 'no_wall_1_person_on_los_x1m': [1.0, 0], 'no_wall_1_person_on_los_x2.5m': [2.5, 0], 'no_wall_1_person_x1.5m_left_0.5m': [1.5, -0.5], 'no_wall_1_person_x1.5m_left_1m': [1.5, -1.0], 'no_wall_1_person_x1.5m_right_0.5m': [1.5, 0.5], 'no_wall_1_person_x1.5m_right_1m': [1.5, 1.0], 'no_wall_1_person_x1m_left_0.75m': [1.0, -0.75], 'no_wall_1_person_x1m_left_1m': [1.0, -1.0], 'no_wall_1_person_x1m_right_0.75m': [1.0, 0.75], 'no_wall_1_person_x1m_right_1m': [1.0, 1.0], 'no_wall_1_person_x2.5m_left_1m': [2.5, -1.0], 'no_wall_1_person_x2.5m_right_1m': [2.5, 1.0], 'no_wall_1_person_x2m_left_0.75m': [2.0, -0.75], 'no_wall_1_person_x2m_right_0.75m': [2.0, 0.75], 'two_walls_2.5m_1_person_behind_router_1.5m': [4.0, 0], 'two_walls_3.25m_1_person_behind_router_1m': [4.25, 0], 'two_walls_3.25m_1_person_midpoint': [1.625, 0], 'two_walls_3.5m_1_person_behind_router_1m': [4.5, 0], 'two_walls_3m_1_person_midpoint': [1.5, 0]}, 'development': ['1.5_meter_los_person', '1_human_in_los_0.5m', '1_meter_Los_person', '1m_off_LOS_left_0.5m', '1m_off_LOS_left_1m', '1m_off_LOS_right_0.5m', '1m_off_los_right_1m', '2_meter_los_person', '2_walls_2.5m_1_person_behind ESPby1m', '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los', '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los', '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint', '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los', '2_walls_4m_1_person_between_midpoint', '2m_off_Los_left_1m', '2m_off_Los_right_1m', '2m_off_los_left_0.5m', '2m_off_los_right_0.5m', 'no_wall_1_person_on_los_x0.75m', 'no_wall_1_person_on_los_x1.25m', 'no_wall_1_person_on_los_x1.75m', 'no_wall_1_person_on_los_x1m', 'no_wall_1_person_on_los_x2.5m', 'no_wall_1_person_x1.5m_left_0.5m', 'no_wall_1_person_x1.5m_left_1m', 'no_wall_1_person_x1.5m_right_0.5m', 'no_wall_1_person_x1.5m_right_1m', 'no_wall_1_person_x1m_left_0.75m', 'no_wall_1_person_x1m_left_1m', 'no_wall_1_person_x1m_right_0.75m', 'no_wall_1_person_x1m_right_1m', 'no_wall_1_person_x2.5m_left_1m', 'no_wall_1_person_x2.5m_right_1m', 'no_wall_1_person_x2m_left_0.75m', 'no_wall_1_person_x2m_right_0.75m', 'two_walls_2.5m_1_person_behind_router_1.5m', 'two_walls_3.25m_1_person_behind_router_1m', 'two_walls_3.25m_1_person_midpoint', 'two_walls_3.5m_1_person_behind_router_1m', 'two_walls_3m_1_person_midpoint'], 'repeats': ['1.5_meter_los_person_repeat', '1m_off_los_right_1m_repeat', '2_meter_los_person_repeat', '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los_repeat', '2_walls_4m_1_person_between_midpoint_repeat', '2m_off_Los_left_1m_repeat'], 'folds': {'fold_1': {'split': {'1.5_meter_los_person': {'packets': 4476, 'split': 'train'}, '1_human_in_los_0.5m': {'packets': 4488, 'split': 'train'}, '1_meter_Los_person': {'packets': 1492, 'split': 'val'}, '1m_off_LOS_left_0.5m': {'packets': 1495, 'split': 'test'}, '1m_off_LOS_left_1m': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_right_0.5m': {'packets': 1492, 'split': 'test'}, '1m_off_los_right_1m': {'packets': 4482, 'split': 'train'}, '2_meter_los_person': {'packets': 4470, 'split': 'train'}, '2_walls_2.5m_1_person_behind ESPby1m': {'packets': 4487, 'split': 'train'}, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': {'packets': 4489, 'split': 'train'}, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': {'packets': 1495, 'split': 'test'}, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': {'packets': 1491, 'split': 'test'}, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': {'packets': 1491, 'split': 'test'}, '2_walls_4m_1_person_between_midpoint': {'packets': 4494, 'split': 'train'}, '2m_off_Los_left_1m': {'packets': 1494, 'split': 'test'}, '2m_off_Los_right_1m': {'packets': 1496, 'split': 'val'}, '2m_off_los_left_0.5m': {'packets': 1494, 'split': 'val'}, '2m_off_los_right_0.5m': {'packets': 1494, 'split': 'val'}, 'no_wall_1_person_on_los_x0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.25m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_on_los_x1.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_on_los_x2.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_right_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x2m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x2m_right_0.75m': {'packets': 2400, 'split': 'val'}, 'two_walls_2.5m_1_person_behind_router_1.5m': {'packets': 1500, 'split': 'val'}, 'two_walls_3.25m_1_person_behind_router_1m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_midpoint': {'packets': 1500, 'split': 'test'}, 'two_walls_3.5m_1_person_behind_router_1m': {'packets': 2400, 'split': 'train'}, 'two_walls_3m_1_person_midpoint': {'packets': 2400, 'split': 'train'}}, 'current': {42: {'val': 0.7466991655528545, 'test': {'1m_off_LOS_left_0.5m': 0.6523025035858154, '1m_off_LOS_right_0.5m': 0.5035436153411865, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': 1.0072354078292847, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': 0.3835340738296509, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': 2.464404821395874, '2m_off_Los_left_1m': 0.9185740351676941, 'no_wall_1_person_on_los_x1.25m': 0.3580451011657715, 'two_walls_3.25m_1_person_midpoint': 0.06789392232894897}}, 123: {'val': 0.7690443210303783, 'test': {'1m_off_LOS_left_0.5m': 0.690818190574646, '1m_off_LOS_right_0.5m': 0.4218222200870514, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': 1.2103230953216553, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': 0.36750900745391846, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': 2.450098991394043, '2m_off_Los_left_1m': 0.789320707321167, 'no_wall_1_person_on_los_x1.25m': 0.3604981005191803, 'two_walls_3.25m_1_person_midpoint': 0.08225631713867188}}, 314: {'val': 0.7459889464080334, 'test': {'1m_off_LOS_left_0.5m': 0.7174301743507385, '1m_off_LOS_right_0.5m': 0.4927988350391388, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': 0.923051118850708, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': 0.41170480847358704, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': 2.079441547393799, '2m_off_Los_left_1m': 0.8663759827613831, 'no_wall_1_person_on_los_x1.25m': 0.3625108301639557, 'two_walls_3.25m_1_person_midpoint': 0.07469190657138824}}}, 'data_sha256': '56cbf2c2b40c98f377a4b3e8e4a4a100f92cb9feb810464a0dcbbd60677fd344'}, 'fold_2': {'split': {'1.5_meter_los_person': {'packets': 4476, 'split': 'train'}, '1_human_in_los_0.5m': {'packets': 4488, 'split': 'train'}, '1_meter_Los_person': {'packets': 1492, 'split': 'test'}, '1m_off_LOS_left_0.5m': {'packets': 4495, 'split': 'train'}, '1m_off_LOS_left_1m': {'packets': 1492, 'split': 'val'}, '1m_off_LOS_right_0.5m': {'packets': 4492, 'split': 'train'}, '1m_off_los_right_1m': {'packets': 1482, 'split': 'val'}, '2_meter_los_person': {'packets': 4470, 'split': 'train'}, '2_walls_2.5m_1_person_behind ESPby1m': {'packets': 1487, 'split': 'val'}, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': {'packets': 4489, 'split': 'train'}, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': {'packets': 4495, 'split': 'train'}, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_between_midpoint': {'packets': 4494, 'split': 'train'}, '2m_off_Los_left_1m': {'packets': 4494, 'split': 'train'}, '2m_off_Los_right_1m': {'packets': 1496, 'split': 'test'}, '2m_off_los_left_0.5m': {'packets': 1494, 'split': 'test'}, '2m_off_los_right_0.5m': {'packets': 1494, 'split': 'test'}, 'no_wall_1_person_on_los_x0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.25m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.75m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_on_los_x1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_on_los_x2.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_left_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_right_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x2.5m_left_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x2.5m_right_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x2m_left_0.75m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_x2m_right_0.75m': {'packets': 2400, 'split': 'test'}, 'two_walls_2.5m_1_person_behind_router_1.5m': {'packets': 1500, 'split': 'test'}, 'two_walls_3.25m_1_person_behind_router_1m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_midpoint': {'packets': 1500, 'split': 'train'}, 'two_walls_3.5m_1_person_behind_router_1m': {'packets': 2400, 'split': 'train'}, 'two_walls_3m_1_person_midpoint': {'packets': 2400, 'split': 'train'}}, 'current': {42: {'val': 1.1144301013814077, 'test': {'1_meter_Los_person': 0.25608253479003906, '2m_off_Los_right_1m': 1.1922950744628906, '2m_off_los_left_0.5m': 0.6772830486297607, '2m_off_los_right_0.5m': 0.896926760673523, 'no_wall_1_person_on_los_x1m': 1.2606714963912964, 'no_wall_1_person_x2.5m_right_1m': 1.0218372344970703, 'no_wall_1_person_x2m_right_0.75m': 0.9075443744659424, 'two_walls_2.5m_1_person_behind_router_1.5m': 0.36986473202705383}}, 123: {'val': 1.1231482194529638, 'test': {'1_meter_Los_person': 0.25322699546813965, '2m_off_Los_right_1m': 1.2385092973709106, '2m_off_los_left_0.5m': 0.7479441165924072, '2m_off_los_right_0.5m': 0.9712544679641724, 'no_wall_1_person_on_los_x1m': 1.2550917863845825, 'no_wall_1_person_x2.5m_right_1m': 1.0329326391220093, 'no_wall_1_person_x2m_right_0.75m': 0.9502174854278564, 'two_walls_2.5m_1_person_behind_router_1.5m': 0.32380685210227966}}, 314: {'val': 1.1036300626065996, 'test': {'1_meter_Los_person': 0.26155152916908264, '2m_off_Los_right_1m': 1.1926891803741455, '2m_off_los_left_0.5m': 0.7251996397972107, '2m_off_los_right_0.5m': 0.9290550947189331, 'no_wall_1_person_on_los_x1m': 1.246582269668579, 'no_wall_1_person_x2.5m_right_1m': 1.0408904552459717, 'no_wall_1_person_x2m_right_0.75m': 0.9053387641906738, 'two_walls_2.5m_1_person_behind_router_1.5m': 0.34004315733909607}}}, 'data_sha256': 'c2f74807a588081ca397997702148199d3f427f5a2bab5ce6162a2f2d2b1a9c1'}, 'fold_3': {'split': {'1.5_meter_los_person': {'packets': 4476, 'split': 'train'}, '1_human_in_los_0.5m': {'packets': 4488, 'split': 'train'}, '1_meter_Los_person': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_left_0.5m': {'packets': 4495, 'split': 'train'}, '1m_off_LOS_left_1m': {'packets': 1492, 'split': 'test'}, '1m_off_LOS_right_0.5m': {'packets': 4492, 'split': 'train'}, '1m_off_los_right_1m': {'packets': 1482, 'split': 'test'}, '2_meter_los_person': {'packets': 1470, 'split': 'val'}, '2_walls_2.5m_1_person_behind ESPby1m': {'packets': 1487, 'split': 'test'}, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': {'packets': 1489, 'split': 'val'}, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': {'packets': 4495, 'split': 'train'}, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_between_midpoint': {'packets': 1494, 'split': 'val'}, '2m_off_Los_left_1m': {'packets': 4494, 'split': 'train'}, '2m_off_Los_right_1m': {'packets': 4496, 'split': 'train'}, '2m_off_los_left_0.5m': {'packets': 4494, 'split': 'train'}, '2m_off_los_right_0.5m': {'packets': 4494, 'split': 'train'}, 'no_wall_1_person_on_los_x0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.25m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.75m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_on_los_x1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_on_los_x2.5m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1.5m_left_0.5m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1.5m_left_1m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1.5m_right_0.5m': {'packets': 1500, 'split': 'val'}, 'no_wall_1_person_x1.5m_right_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_left_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x1m_right_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x2.5m_left_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x2.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2m_left_0.75m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_x2m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'two_walls_2.5m_1_person_behind_router_1.5m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_behind_router_1m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_midpoint': {'packets': 1500, 'split': 'train'}, 'two_walls_3.5m_1_person_behind_router_1m': {'packets': 2400, 'split': 'val'}, 'two_walls_3m_1_person_midpoint': {'packets': 2400, 'split': 'train'}}, 'current': {42: {'val': 0.6931973975151777, 'test': {'1m_off_LOS_left_1m': 0.8207272887229919, '1m_off_los_right_1m': 1.055274486541748, '2_walls_2.5m_1_person_behind ESPby1m': 4.163684368133545, 'no_wall_1_person_on_los_x1.75m': 0.3170345425605774, 'no_wall_1_person_x1.5m_right_1m': 0.48084723949432373, 'no_wall_1_person_x1m_left_1m': 0.7793841361999512, 'no_wall_1_person_x1m_right_1m': 1.1632426977157593, 'no_wall_1_person_x2.5m_left_1m': 1.2144699096679688, 'no_wall_1_person_x2m_left_0.75m': 0.5215091705322266}}, 123: {'val': 0.6804662384092808, 'test': {'1m_off_LOS_left_1m': 0.8655231595039368, '1m_off_los_right_1m': 0.8726287484169006, '2_walls_2.5m_1_person_behind ESPby1m': 3.8938796520233154, 'no_wall_1_person_on_los_x1.75m': 0.3452204465866089, 'no_wall_1_person_x1.5m_right_1m': 0.4082147479057312, 'no_wall_1_person_x1m_left_1m': 0.8485497832298279, 'no_wall_1_person_x1m_right_1m': 1.0257477760314941, 'no_wall_1_person_x2.5m_left_1m': 1.3227674961090088, 'no_wall_1_person_x2m_left_0.75m': 0.5505629181861877}}, 314: {'val': 0.6666196174919605, 'test': {'1m_off_LOS_left_1m': 0.7436093688011169, '1m_off_los_right_1m': 0.9148960709571838, '2_walls_2.5m_1_person_behind ESPby1m': 3.8301477432250977, 'no_wall_1_person_on_los_x1.75m': 0.36070266366004944, 'no_wall_1_person_x1.5m_right_1m': 0.4413476586341858, 'no_wall_1_person_x1m_left_1m': 0.786289632320404, 'no_wall_1_person_x1m_right_1m': 0.981619656085968, 'no_wall_1_person_x2.5m_left_1m': 1.4454478025436401, 'no_wall_1_person_x2m_left_0.75m': 0.5848421454429626}}}, 'data_sha256': 'ff2655352a54669558440ebd89987c21d8e2bad0a75bbc1c4654a87de29574f1'}, 'fold_4': {'split': {'1.5_meter_los_person': {'packets': 1476, 'split': 'val'}, '1_human_in_los_0.5m': {'packets': 1488, 'split': 'val'}, '1_meter_Los_person': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_left_0.5m': {'packets': 4495, 'split': 'train'}, '1m_off_LOS_left_1m': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_right_0.5m': {'packets': 4492, 'split': 'train'}, '1m_off_los_right_1m': {'packets': 4482, 'split': 'train'}, '2_meter_los_person': {'packets': 1470, 'split': 'test'}, '2_walls_2.5m_1_person_behind ESPby1m': {'packets': 4487, 'split': 'train'}, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': {'packets': 1489, 'split': 'test'}, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': {'packets': 4495, 'split': 'train'}, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': {'packets': 4491, 'split': 'train'}, '2_walls_4m_1_person_between_midpoint': {'packets': 1494, 'split': 'test'}, '2m_off_Los_left_1m': {'packets': 4494, 'split': 'train'}, '2m_off_Los_right_1m': {'packets': 4496, 'split': 'train'}, '2m_off_los_left_0.5m': {'packets': 4494, 'split': 'train'}, '2m_off_los_right_0.5m': {'packets': 4494, 'split': 'train'}, 'no_wall_1_person_on_los_x0.75m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_on_los_x1.25m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_on_los_x2.5m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1.5m_left_0.5m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1.5m_left_1m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1.5m_right_0.5m': {'packets': 1500, 'split': 'test'}, 'no_wall_1_person_x1.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_left_0.75m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_x1m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_right_0.75m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_x1m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x2m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'two_walls_2.5m_1_person_behind_router_1.5m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_behind_router_1m': {'packets': 1500, 'split': 'val'}, 'two_walls_3.25m_1_person_midpoint': {'packets': 1500, 'split': 'train'}, 'two_walls_3.5m_1_person_behind_router_1m': {'packets': 2400, 'split': 'test'}, 'two_walls_3m_1_person_midpoint': {'packets': 2400, 'split': 'val'}}, 'current': {42: {'val': 0.8255455068179539, 'test': {'2_meter_los_person': 0.49680662155151367, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': 0.6919661164283752, '2_walls_4m_1_person_between_midpoint': 1.7349936962127686, 'no_wall_1_person_on_los_x2.5m': 1.1955379247665405, 'no_wall_1_person_x1.5m_left_0.5m': 0.4963946044445038, 'no_wall_1_person_x1.5m_left_1m': 0.4762519896030426, 'no_wall_1_person_x1.5m_right_0.5m': 0.41184452176094055, 'two_walls_3.5m_1_person_behind_router_1m': 1.2230238914489746}}, 123: {'val': 0.8269749635032245, 'test': {'2_meter_los_person': 0.5221271514892578, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': 0.7700662016868591, '2_walls_4m_1_person_between_midpoint': 1.6988173723220825, 'no_wall_1_person_on_los_x2.5m': 1.1453633308410645, 'no_wall_1_person_x1.5m_left_0.5m': 0.450030118227005, 'no_wall_1_person_x1.5m_left_1m': 0.49600520730018616, 'no_wall_1_person_x1.5m_right_0.5m': 0.3735640347003937, 'two_walls_3.5m_1_person_behind_router_1m': 1.5346786975860596}}, 314: {'val': 0.8373823197824615, 'test': {'2_meter_los_person': 0.5423778891563416, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': 0.8427553176879883, '2_walls_4m_1_person_between_midpoint': 1.3744196891784668, 'no_wall_1_person_on_los_x2.5m': 1.1400675773620605, 'no_wall_1_person_x1.5m_left_0.5m': 0.39103567600250244, 'no_wall_1_person_x1.5m_left_1m': 0.45519784092903137, 'no_wall_1_person_x1.5m_right_0.5m': 0.38619816303253174, 'two_walls_3.5m_1_person_behind_router_1m': 1.790583848953247}}}, 'data_sha256': '27559dea4c33c4bedfa0ef7f5d9c6c38466e18ba49267900649b9ebea4704b3c'}, 'fold_5': {'split': {'1.5_meter_los_person': {'packets': 1476, 'split': 'test'}, '1_human_in_los_0.5m': {'packets': 1488, 'split': 'test'}, '1_meter_Los_person': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_left_0.5m': {'packets': 1495, 'split': 'val'}, '1m_off_LOS_left_1m': {'packets': 4492, 'split': 'train'}, '1m_off_LOS_right_0.5m': {'packets': 1492, 'split': 'val'}, '1m_off_los_right_1m': {'packets': 4482, 'split': 'train'}, '2_meter_los_person': {'packets': 4470, 'split': 'train'}, '2_walls_2.5m_1_person_behind ESPby1m': {'packets': 4487, 'split': 'train'}, '2_walls_2.5m_1_person_behind_router_by_2m_collinear_with_los': {'packets': 4489, 'split': 'train'}, '2_walls_2.5m_1_person_behind_the_router_by_1m_collinear_with_los': {'packets': 1495, 'split': 'val'}, '2_walls_2.5m_1_person_between_router_and_esp_in_los_midpoint': {'packets': 1491, 'split': 'val'}, '2_walls_4m_1_person_behind_router_by_1m_collinear_with_los': {'packets': 1491, 'split': 'val'}, '2_walls_4m_1_person_between_midpoint': {'packets': 4494, 'split': 'train'}, '2m_off_Los_left_1m': {'packets': 1494, 'split': 'val'}, '2m_off_Los_right_1m': {'packets': 4496, 'split': 'train'}, '2m_off_los_left_0.5m': {'packets': 4494, 'split': 'train'}, '2m_off_los_right_0.5m': {'packets': 4494, 'split': 'train'}, 'no_wall_1_person_on_los_x0.75m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_on_los_x1.25m': {'packets': 2400, 'split': 'val'}, 'no_wall_1_person_on_los_x1.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_on_los_x1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_on_los_x2.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_0.5m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_left_0.75m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_x1m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x1m_right_0.75m': {'packets': 2400, 'split': 'test'}, 'no_wall_1_person_x1m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_left_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2.5m_right_1m': {'packets': 1500, 'split': 'train'}, 'no_wall_1_person_x2m_left_0.75m': {'packets': 2400, 'split': 'train'}, 'no_wall_1_person_x2m_right_0.75m': {'packets': 2400, 'split': 'train'}, 'two_walls_2.5m_1_person_behind_router_1.5m': {'packets': 1500, 'split': 'train'}, 'two_walls_3.25m_1_person_behind_router_1m': {'packets': 1500, 'split': 'test'}, 'two_walls_3.25m_1_person_midpoint': {'packets': 1500, 'split': 'val'}, 'two_walls_3.5m_1_person_behind_router_1m': {'packets': 2400, 'split': 'train'}, 'two_walls_3m_1_person_midpoint': {'packets': 2400, 'split': 'test'}}, 'current': {42: {'val': 0.7869975566864014, 'test': {'1.5_meter_los_person': 0.34841057658195496, '1_human_in_los_0.5m': 1.3699209690093994, 'no_wall_1_person_on_los_x0.75m': 0.7515928149223328, 'no_wall_1_person_x1m_left_0.75m': 0.6435409784317017, 'no_wall_1_person_x1m_right_0.75m': 0.6838490962982178, 'two_walls_3.25m_1_person_behind_router_1m': 0.25268110632896423, 'two_walls_3m_1_person_midpoint': 0.3422497808933258}}, 123: {'val': 0.7320957668125629, 'test': {'1.5_meter_los_person': 0.33913904428482056, '1_human_in_los_0.5m': 1.3586499691009521, 'no_wall_1_person_on_los_x0.75m': 0.70306795835495, 'no_wall_1_person_x1m_left_0.75m': 0.5882057547569275, 'no_wall_1_person_x1m_right_0.75m': 0.6882994771003723, 'two_walls_3.25m_1_person_behind_router_1m': 0.2541011571884155, 'two_walls_3m_1_person_midpoint': 0.3528931140899658}}, 314: {'val': 0.7894107941538095, 'test': {'1.5_meter_los_person': 0.29486507177352905, '1_human_in_los_0.5m': 1.1370779275894165, 'no_wall_1_person_on_los_x0.75m': 0.6620199680328369, 'no_wall_1_person_x1m_left_0.75m': 0.6813783049583435, 'no_wall_1_person_x1m_right_0.75m': 0.5864339470863342, 'two_walls_3.25m_1_person_behind_router_1m': 0.37398001551628113, 'two_walls_3m_1_person_midpoint': 0.26302051544189453}}}, 'data_sha256': '6a330a0157f3d8ed1b5d734dd059ed188fead2bb040a9aa3037eebe4ed7d1977'}}}

import csv
import hashlib
import json
import shutil
import subprocess
import sys
from pathlib import Path

import numpy as np
import torch
from threadpoolctl import threadpool_limits

scratch = Path("/kaggle/temp/throughwi-model-comparison")
out = Path("/kaggle/working/throughwi-model-comparison")
out.mkdir(parents=True, exist_ok=True)
scratch.mkdir(parents=True, exist_ok=True)
repo = scratch / "repo"
if not (repo / ".git").exists():
    subprocess.run(
        ["git", "clone", "--depth", "1",
         "https://github.com/Hussein-Yaacoub/throughwi-csi-localization-finalized.git", str(repo)],
        check=True,
    )
commit = subprocess.check_output(["git", "-C", str(repo), "rev-parse", "HEAD"], text=True).strip()
if commit != PROTOCOL["commit"]:
    raise ValueError(f"Expected {PROTOCOL['commit']}, found {commit}.")
checkpoint = repo / "kaggle" / "haloc-best.pt"
assert hashlib.sha256(checkpoint.read_bytes()).hexdigest() == PROTOCOL["source_checkpoint_sha256"]
assert torch.cuda.is_available(), "Select a Kaggle GPU."
torch.backends.cudnn.allow_tf32 = False
torch.backends.cuda.matmul.allow_tf32 = False
code = repo / "experiments" / "residential_cnn"
sys.path.insert(0, str(code))
from prepare_captures import csi_features
from recover_train import data_digest, load_data
from cross_evaluate import read_checkpoint, score

seeds = (42, 123, 314)
# Preserve the original model and trainer. Keep the phase channel zero after
# packet augmentation as well as at validation and test time.
wrapper = scratch / "train_amplitude.py"
wrapper.write_text(
    "import sys\n"
    f"sys.path.insert(0, {str(code)!r})\n"
    "import argparse\n"
    "from pathlib import Path\n"
    "import recover_train as original\n"
    "class AmplitudePackets(original.Packets):\n"
    "    def __getitem__(self, index):\n"
    "        x, y = super().__getitem__(index)\n"
    "        x = x.clone()\n"
    "        x[1].zero_()\n"
    "        return x, y\n"
    "original.Packets = AmplitudePackets\n"
    "parser = argparse.ArgumentParser()\n"
    "parser.add_argument('data', type=Path)\n"
    "parser.add_argument('--out', type=Path, required=True)\n"
    "parser.add_argument('--init', type=Path, required=True)\n"
    "parser.add_argument('--seed', type=int, required=True)\n"
    "parser.add_argument('--resume', action='store_true')\n"
    "parser.set_defaults(epochs=30, batch_size=512, threads=2, device='cuda', max_epochs_this_call=None)\n"
    "original.run(parser.parse_args())\n"
)

data_dir = repo / "data"
def read_packets(path):
    features = []
    with path.open(encoding="utf-8-sig", newline="") as stream:
        for row in csv.DictReader(stream):
            try:
                if float(row["csi_len"]) != 256:
                    continue
                values = np.fromstring(row["csi_raw"].strip().strip("[]").replace(",", " "), sep=" ", dtype=np.float32)
                if values.size == 256 and np.isfinite(values).all():
                    features.append(csi_features(values, "iq"))
            except (ValueError, TypeError, AttributeError):
                continue
    if not features:
        raise ValueError(f"No usable CSI packets in {path.name}")
    return np.asarray(features, dtype=np.float32)

def ids(features):
    return [hashlib.sha256(row.tobytes()).digest() for row in features]

records = {}
for name in PROTOCOL["development"] + PROTOCOL["repeats"]:
    raw = read_packets(data_dir / f"{name}.csv")
    aug_path = data_dir / f"{name}_augmented.csv"
    aug = read_packets(aug_path) if aug_path.exists() else np.empty((0, 256), dtype=np.float32)
    records[name] = raw, aug, ids(raw), ids(aug)
protected_repeat = {v for name in PROTOCOL["repeats"] for v in records[name][2]}

def make_dataset(fold):
    roles = {role: sorted(n for n, v in fold["split"].items() if v["split"] == role)
             for role in ("train", "val", "test")}
    protected_test = {v for n in roles["test"] for v in records[n][2]}
    protected = protected_repeat | protected_test | {v for n in roles["val"] for v in records[n][2]}
    arrays = {k: [] for k in ("X", "Y", "recording", "split")}
    for split, names in roles.items():
        for name in names:
            raw, aug, raw_ids, aug_ids = records[name]
            x = raw
            if split == "train":
                x = np.concatenate([raw, aug])[[v not in protected for v in raw_ids + aug_ids]]
            if split == "val":
                x = raw[[v not in protected_test and v not in protected_repeat for v in raw_ids]]
            assert len(x) == fold["split"][name]["packets"], (name, len(x))
            arrays["X"].append(x)
            arrays["Y"].append(np.tile(np.asarray((*PROTOCOL["xy"][name], 1), dtype=np.float32), (len(x), 1)))
            arrays["recording"].append(np.repeat(name, len(x)))
            arrays["split"].append(np.repeat(split, len(x)))
    data = {key: np.concatenate(value) for key, value in arrays.items()}
    data["byte_order"] = np.array("iq")
    assert data_digest(data) == fold["data_sha256"], "The reconstructed inputs differ from the previous experiment."
    return data

def recording_errors(predict, data, split):
    mask = data["split"] == split
    e = np.linalg.norm(predict - data["Y"][mask, :2], axis=1)
    names = data["recording"][mask]
    return {str(n): float(e[names == n].mean()) for n in np.unique(names)}

def ridge_predictors(data):
    train = data["split"] == "train"
    x = np.log1p(data["X"][:, :128]).astype(np.float64)
    mu, sd = x[train].mean(axis=0), x[train].std(axis=0)
    x = (x - mu) / np.maximum(sd, 1e-8)
    y = data["Y"][train, :2].astype(np.float64)
    center = y.mean(axis=0)
    gram = x[train].T @ x[train] / train.sum()
    rhs = x[train].T @ (y - center) / train.sum()
    with threadpool_limits(limits=2, user_api="blas"):
        for strength in (0.001, 0.01, 0.1, 1.0):
            weights = np.linalg.solve(gram + strength * np.eye(gram.shape[0]), rhs)
            yield strength, weights, center, mu, sd

def mean(values):
    return float(np.mean(list(values)))

def save_json(path, value):
    path.write_text(json.dumps(value, indent=2, sort_keys=True) + "\n")

print("Data and checkpoint verified:", commit, flush=True)
results = {}
for tag, fold in PROTOCOL["folds"].items():
    print("START", tag, flush=True)
    data = make_dataset(fold)
    ridge_candidates = []
    for strength, weights, center, mu, sd in ridge_predictors(data):
        hold = data["split"] == "val"
        xx = np.log1p(data["X"][hold, :128]).astype(np.float64)
        yy = ((xx - mu) / np.maximum(sd, 1e-8)) @ weights + center
        ridge_candidates.append((mean(recording_errors(yy, data, "val").values()), strength, weights, center, mu, sd))
    ridge_val, strength, weights, center, mu, sd = min(ridge_candidates, key=lambda item: item[0])
    np.savez_compressed(out / f"{tag}_ridge.npz", weights=weights, center=center, input_mean=mu, input_sd=sd)
    original_test = data["X"][data["split"] == "test", :128]
    data["X"][:, 128:] = 0
    fold_path = scratch / (tag + "_amplitude.npz")
    np.savez(fold_path, **data)
    result = {"current": fold["current"], "amplitude": {}, "ridge_validation": ridge_val, "ridge_strength": strength}
    for seed in seeds:
        run = scratch / f"{tag}_seed_{seed}"
        output = out / f"{tag}_seed_{seed}"
        output.mkdir(exist_ok=True)
        print("TRAIN amplitude-only", tag, seed, flush=True)
        cmd = [sys.executable, "-u", str(wrapper), str(fold_path), "--out", str(run),
               "--init", str(checkpoint), "--seed", str(seed)]
        if (run / "last.pt").exists():
            cmd.append("--resume")
        if not (run / "metrics.json").exists():
            subprocess.run(cmd, check=True)
        run_metrics = json.loads((run / "metrics.json").read_text())
        result["amplitude"][seed] = {"val": run_metrics["val"]["recording_balanced_mean_distance_m"]}
        for filename in ("best.pt", "metrics.json", "splits.json"):
            shutil.copy2(run / filename, output)
    validation = {
        "current": mean(v["val"] for v in result["current"].values()),
        "amplitude": mean(v["val"] for v in result["amplitude"].values()),
        "ridge": ridge_val,
    }
    winner = min(validation, key=validation.get)
    result.update({"validation": validation, "selected": winner})
    print("VALIDATION SELECTION", tag, validation, winner, flush=True)
    for seed in seeds:
        output = out / f"{tag}_seed_{seed}"
        metrics, pred, labels, names = score(
            read_checkpoint(output / "best.pt"), data, torch.device("cuda"), use_amp=False
        )
        result["amplitude"][seed]["test"] = metrics["per_recording_mean_horizontal_distance_m"]
        np.savez_compressed(output / "test_full_precision.npz", predictions=pred, labels=labels, recording=names)
    xx = np.log1p(original_test).astype(np.float64)
    ridge_test = ((xx - mu) / np.maximum(sd, 1e-8)) @ weights + center
    result["ridge_test"] = recording_errors(ridge_test, data, "test")
    results[tag] = result
    save_json(out / "fold_results.json", results)
    save_json(out / "protocol.json", {"repository_commit": commit, "seeds": seeds, "exploratory_test_reuse": True})
    fold_path.unlink()

summary = {}
for model in ("current", "amplitude", "ridge", "validation_selected"):
    seed_means = []
    for seed in seeds:
        errors = []
        for fold in results.values():
            use = fold["selected"] if model == "validation_selected" else model
            if use == "ridge":
                values = fold["ridge_test"]
            else:
                variant = fold[use]
                values = variant.get(seed, variant.get(str(seed)))["test"]
            errors.extend(values.values())
        assert len(errors) == 40
        seed_means.append(mean(errors))
    summary[model] = {"recording_balanced_mean_m": mean(seed_means),
                      "seed_sd_m": float(np.std(seed_means, ddof=1)), "by_seed_m": seed_means}
save_json(out / "summary.json", summary)
shutil.make_archive("/kaggle/working/throughwi-model-comparison", "zip", out)
print("FINAL EXPLORATORY COMPARISON", json.dumps(summary), flush=True)
```
